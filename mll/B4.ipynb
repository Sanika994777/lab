```python

# Gradient Descent Algorithm

# Define the function
def function(x):
    return (x + 3) ** 2

# Define the derivative (gradient)
def gradient(x):
    return 2 * (x + 3)

# Initialize parameters
x = 2
learning_rate = 0.1
iterations = 100

print("Initial x =", x)
print("Initial y =", function(x))

# Perform gradient descent
for i in range(iterations):
    x = x - learning_rate * gradient(x)

# Display the result
print("\nMinimum point x =", round(x, 4))
print("Minimum value y =", round(function(x), 4))

#TC:O(N) & SC:O(1)
```
